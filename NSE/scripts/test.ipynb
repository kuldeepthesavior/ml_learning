```python
from openchart import NSEData
import datetime

# Initialize the NSEData class
nse = NSEData()


```

```python
nse.historical(symbol='NIFTY', segment='FO', interval='5m')
```

```python
# Download master data for NSE and NFO
# nse.download()

# Define the start and end dates (last 30 days)
end_date = datetime.datetime.now()
start_date = end_date - datetime.timedelta(days=740)
```

```python
symbols = nse.search('NIFTY', 'FO')
symbols
```

```python
# Fetch 5-minute historical data for RELIANCE
data = nse.historical(
    symbol='NIFTY2620324900PE', #NIFTY, BANKNIFTY, FINNIFTY, NIFTY2620324900PE
    start=start_date,
    end=end_date,
    interval='5m',
    segment='FO'
)

# Display the fetched data
if not data.empty:
    print("5-minute historical data for RELIANCE (Last 30 days):")
    print(data)
else:
    print("No data available for RELIANCE for the specified time period.")
```
