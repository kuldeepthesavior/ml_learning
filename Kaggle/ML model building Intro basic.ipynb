### Notes on Specific Variables
Rooms: Number of rooms

Price: Price in dollars

Method: S - property sold; SP - property sold prior; PI - property passed in; PN - sold prior not disclosed; SN - sold not disclosed; NB - no bid; VB - vendor bid; W - withdrawn prior to auction; SA - sold after auction; SS - sold after auction price not disclosed. N/A - price or highest bid not available.

Type: br - bedroom(s); h - house,cottage,villa, semi,terrace; u - unit, duplex; t - townhouse; dev site - development site; o res - other residential.

SellerG: Real Estate Agent

Date: Date sold

Distance: Distance from CBD

Regionname: General Region (West, North West, North, North east …etc)

Propertycount: Number of properties that exist in the suburb.

Bedroom2 : Scraped # of Bedrooms (from different source)

Bathroom: Number of Bathrooms

Car: Number of carspots

Landsize: Land Size

BuildingArea: Building Size

CouncilArea: Governing council for the area

```python
# Converting the categorical variables into numerical

from sklearn.preprocessing import LabelEncoder

lblEncoder = LabelEncoder()
```

```python
for col in df.columns.to_list():
    if df[col].dtypes == 'object':
        print(col)
        df[f'{col}_num'] = lblEncoder.fit_transform(df[col])
        print(dict(zip(lblEncoder.classes_, lblEncoder.transform(lblEncoder.classes_))))
```

```python
df.head()
```

```python
import datetime
df['PropAge'] = df['YearBuilt'] - datetime.datetime.today().year
```

```python
df.columns
```

```python
features = [ 'Rooms', 'Type_num', 'Distance', 'Bedroom2', 'Bathroom', 'Car',
       'Landsize', 'BuildingArea', 'PropAge', 'Propertycount','Suburb_num',
        'Method_num', 'Regionname_num' ]

df1 = df[features]
df1.head()
```

```python
df1['Car'].fillna(df1['Car'].mean(), inplace=True)
df1['BuildingArea'].fillna(df1['BuildingArea'].median(), inplace=True)
df1['PropAge'].fillna(df1['PropAge'].median(), inplace=True)
df1.isna().mean() * 100
```

```python
X = df1.copy()
y = df.Price
```

## Starting with decision tree menthod and train, test, and validate model

```python
from sklearn.model_selection import train_test_split

train_X, test_X, train_y, test_y = train_test_split(X, y, random_state=10)

from sklearn.tree import DecisionTreeRegressor
dtr = DecisionTreeRegressor()


dtr.fit(train_X, train_y)

dtr

```

```python
predictions_y = dtr.predict(test_X)
```

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(test_y, predictions_y)
```

```python
mae
```

```python
## Checking decision tree property

print(f'features Importance: {dict(zip(dtr.feature_names_in_, dtr.feature_importances_))}')
```

```python
def test_tree_depth(max_leaf_nodes, train_X, test_X, train_y, test_y):
    dtr = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=10)
    dtr.fit(train_X, train_y)
    predictions = dtr.predict(test_X)
    mae = mean_absolute_error(test_y, predictions)
    print(f'Mean Absolute error for max leaf node {max_leaf_nodes} is : {mae}')
    return mae
```

```python
b_leaf = {}
for n_leaf in [10,50,100,500,1000]:
    mae = test_tree_depth(max_leaf_nodes=n_leaf, 
                          train_X=train_X, test_X=test_X, train_y=train_y, test_y=test_y)
    b_leaf[n_leaf] = mae
best_max_leaf = min(b_leaf, key = b_leaf.get) 
```

```python
best_dtr = DecisionTreeRegressor(max_leaf_nodes=best_max_leaf, random_state=1)
best_dtr.fit(X, y)

predictions = best_dtr.predict(X)
mae = mean_absolute_error(predictions, y)
mae
```

## Now switching to Random Forest regressor for better accuracy and minimise the error

```python
from sklearn.ensemble import RandomForestRegressor

rfr = RandomForestRegressor(random_state=10)

rfr.fit(train_X, train_y)
```

```python
predictions = rfr.predict(test_X)
mae = mean_absolute_error(predictions, test_y)
mae
```
